# 01 - EDA

Exploratory look at class balance, image sizes, and format across SID_Set / CIFAKE / WildFake before training. Run after populating `data/raw/` per `data/README.md`.

```python
import glob, os
from PIL import Image
import pandas as pd

DATA_ROOT = 'data/raw'
```

```python
rows = []
for dataset in os.listdir(DATA_ROOT):
    d = os.path.join(DATA_ROOT, dataset)
    if not os.path.isdir(d):
        continue
    n = sum(len(files) for _, _, files in os.walk(d))
    rows.append({'dataset': dataset, 'n_files': n})
pd.DataFrame(rows)
```
